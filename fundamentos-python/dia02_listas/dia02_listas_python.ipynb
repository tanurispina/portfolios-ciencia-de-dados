#Desafio 1: Auditoria e Fusão de Logs (copy e extend)

Na engenharia de dados, você recebe lotes de informações em horários diferentes e precisa unificá-los sem perder o histórico original para caso a auditoria peça.

Crie duas listas de relatórios técnicos (IDs dos chamados): chamados_manha = [101, 102, 103] e
chamados_tarde = [104, 105].

Faça um backup seguro da lista da manhã usando o método .copy() e guarde na variável backup_manha.

Use o método .extend() para injetar os chamados da tarde dentro da lista chamados_manha original.

Imprima o backup_manha (que deve ter só 3 itens) e a lista chamados_manha atualizada (que deve ter 5 itens).

```python
chamados_manha=[101,102,103]
chamados_tarde=[104,105]

backup_manha=chamados_manha.copy()
print(backup_manha)
chamados_manha.extend(chamados_tarde)
print(chamados_manha)
```

##Desafio 2: Triagem de Anomalias (sort(reverse=True))

Quando um sistema de monitoramento acusa aquecimento, a primeira coisa que o cientista de dados faz é isolar os picos de temperatura.

Crie uma lista com registros de temperatura do processador: temperaturas = [45.5, 89.2, 34.1, 99.9, 60.0, 75.5]

Ordene essa lista do maior para o menor usando .sort(reverse=True).

Usando o fatiamento (slicing) que você já domina, imprima apenas o Top 3 de maiores temperaturas (os três primeiros itens da lista ordenada).

```python
temperaturas = [45.5, 89.2, 34.1, 99.9, 60.0, 75.5]
temperaturas.sort(reverse=True)
print(temperaturas)
lista_fatiada=temperaturas[:3]
print(lista_fatiada)

```

## Desafio 3: A Matriz de Dados

Bancos de dados tabulares são lidos no Python como listas dentro de listas. Cada sub-lista é uma "linha".

Crie uma matriz simulando o status de 3 roteadores na rede: tabela_roteadores = [ ["RT-01", "Ativo"], ["RT-02", "Falha"], ["RT-03", "Ativo"] ]

O seu objetivo é acessar especificamente a palavra "Falha" do segundo roteador.

Lembre-se da regra de ouro: primeiro você acessa o índice da linha (a sub-lista do meio), e depois o índice da coluna (a segunda posição dessa sub-lista).

Imprima esse status isolado no console.

```python
tabela_roteadores = [ ["RT-01", "Ativo"], ["RT-02", "Falha"], ["RT-03", "Ativo"] ]
status_tabela=tabela_roteadores[1][1]
print(status_tabela)
```
